```python
# ---- Custom CrewAI Tool for Web Search ----
class TavilySearchTool(BaseTool):
    name: str = "Web Search"
    description: str = "Search the web for recent information."

    def _run(self, query: str):
        url = "https://api.tavily.com/search"

        payload = {
            "api_key": TAVILY_API_KEY,
            "query": query,
            "max_results": 3
        }

        response = requests.post(url, json=payload, timeout=30)
        response.raise_for_status()
        data = response.json()
        results = data.get("results", [])

        if not results:
            return "No search results returned."

        formatted_results = []
        for r in results:
            formatted_results.append(f"{r.get('title', 'Untitled')} - {r.get('url', '')}")

        return "\n".join(formatted_results)


search_tool = TavilySearchTool()
```

Test pdf search tool of crewai_tools

PDF search tool for static, domain-specific content 

Defaults to using OpenAI's API for embeddings and search, which requires the OPENAI_API_KEY environment variable to be set. Since you are using Azure OpenAI, we can pass our custom Azure LLM configuration directly into the tool config so it does not look for a standard OPENAI_API_KEY.

We need to add controlled chunk into the same vector database, without relying on unsupported config keys. Issue is that for bigger pdf documents , the chunking strategy fails as PDFSearchTool does not publicly expose any custom chunker size and tries to load the entire document at once causing RateLimitError.Hence enforce a manually crafter chunker to split the pdf first and add it iteratively into the vector store . 

```python
os.environ["AZURE_OPENAI_API_KEY"] = AZURE_OPENAI_API_KEY
os.environ["AZURE_OPENAI_API_BASE"] = AZURE_OPENAI_ENDPOINT.rstrip("/")
os.environ["AZURE_OPENAI_API_TYPE"] = "azure"
os.environ["AZURE_OPENAI_API_VERSION"] = AZURE_OPENAI_API_VERSION
os.environ["AZURE_OPENAI_DEPLOYMENT"] = AZURE_OPENAI_EMBEDDING_DEPLOYMENT

pdf_path = "/home/subhroy468/router-retriever/venv/attentionPaper.pdf"

# Create a PDFSearchTool instance with only valid config keys
pdf_tool = PDFSearchTool(
    pdf=pdf_path,
    config={
        "embedding_model": {
            "provider": "azure",
            "config": {
                "api_key": AZURE_OPENAI_API_KEY,
                "api_base": AZURE_OPENAI_ENDPOINT.rstrip("/"),
                "api_type": "azure",
                "api_version": AZURE_OPENAI_API_VERSION,
                "deployment_id": AZURE_OPENAI_EMBEDDING_DEPLOYMENT,
                "model_name": AZURE_OPENAI_EMBEDDING_DEPLOYMENT,
            },
        },
        "vectordb": {
            "provider": "chromadb",
            "config": {}
        },
    }
)

# Use the lower-level loader to read the PDF content
loader = PDFLoader()
doc = loader.load(SourceContent(pdf_path))

# Custom chunking strategy
chunker = TextChunker(chunk_size=1200, chunk_overlap=150)

chunks = chunker.chunk(doc.content)

print(f"Total chunks: {len(chunks)}")

# Add chunks one by one to the same vector store
for idx, chunk in enumerate(chunks):
    pdf_tool.add(
        chunk,
        data_type=DataType.TEXT,
        metadata={
            "source_pdf": pdf_path,
            "chunk_index": idx,
            "source_type": "attention_paper"
        }
    )

print("PDF chunks added to vector store.")
```

```python
# pip install pypdf chromadb openai crewai
import time
import chromadb
from pypdf import PdfReader
from openai import AzureOpenAI, RateLimitError
from crewai import Agent, Task, Crew, LLM
from crewai.tools import tool

PDF_PATH = "/home/subhroy468/router-retriever/venv/attentionPaper.pdf"
CHROMA_DIR = "./chroma_store"
COLLECTION = "attention_paper"
BATCH_SIZE = 16  # stays under Azure's per-request input limit

client = AzureOpenAI(
    api_key=AZURE_OPENAI_API_KEY,
    azure_endpoint=AZURE_OPENAI_ENDPOINT,
    api_version=AZURE_OPENAI_API_VERSION,
)

# ---------- 1) Chunk, page by page ----------
def chunk_text(text, size=1000, overlap=150):
    chunks, start = [], 0
    while start < len(text):
        chunks.append(text[start:start + size])
        start += size - overlap
    return chunks

def load_chunks(pdf_path):
    reader = PdfReader(pdf_path)
    docs, metas, ids = [], [], []
    for page_no, page in enumerate(reader.pages, start=1):
        text = (page.extract_text() or "").strip()
        for i, chunk in enumerate(chunk_text(text)):
            if chunk.strip():
                docs.append(chunk)
                metas.append({"page": page_no})
                ids.append(f"p{page_no}-c{i}")
    return docs, metas, ids

# ---------- 2) Embed in small batches ----------
def embed_batch(texts, retries=5):
    for attempt in range(retries):
        try:
            resp = client.embeddings.create(
                model=AZURE_OPENAI_EMBEDDING_DEPLOYMENT,  # deployment name
                input=texts,
            )
            return [d.embedding for d in resp.data]
        except RateLimitError:
            time.sleep(2 ** attempt)
    raise RuntimeError("Embedding failed after retries")

# ---------- 3) Store in Chroma (persistent, skip if already built) ----------
chroma = chromadb.PersistentClient(path=CHROMA_DIR)
collection = chroma.get_or_create_collection(COLLECTION)

def build_index():
    if collection.count() > 0:
        print(f"Index exists with {collection.count()} chunks, skipping.")
        return
    docs, metas, ids = load_chunks(PDF_PATH)
    print(f"{len(docs)} chunks to embed")
    for i in range(0, len(docs), BATCH_SIZE):
        d, m, idx = docs[i:i+BATCH_SIZE], metas[i:i+BATCH_SIZE], ids[i:i+BATCH_SIZE]
        collection.add(documents=d, metadatas=m, ids=idx, embeddings=embed_batch(d))
        print(f"  embedded {min(i+BATCH_SIZE, len(docs))}/{len(docs)}")

build_index()

# ---------- Custom tool for the agent ----------
@tool("Search Attention Paper")
def search_paper(query: str) -> str:
    """Semantic search over the 'Attention Is All You Need' PDF. Returns the most relevant passages with page numbers."""
    q_emb = embed_batch([query])[0]
    res = collection.query(query_embeddings=[q_emb], n_results=4)
    return "\n\n".join(
        f"[page {m['page']}] {doc}"
        for doc, m in zip(res["documents"][0], res["metadatas"][0])
    )
```

```python
# --------------------------------------------------
# Agents for serial and parallel execution  exercise 2
# --------------------------------------------------
researcher = Agent(
    role="AI Researcher",
    goal="Find the latest advancements in AI for healthcare",
    backstory=(
        "You are an expert in artificial intelligence and stay updated "
        "with the latest research trends in healthcare."
    ),
    verbose=True,
    allow_delegation=False,
    llm=llm,
    max_iter=5,
    tools=[search_tool]
)

writer = Agent(
    role="Technical Writer",
    goal="Summarize research into an executive report",
    backstory=(
        "You are an experienced technical writer with expertise in "
        "summarizing healthcare research for executives."
    ),
    verbose=True,
    allow_delegation=False,
    llm=llm
)
```

```python
# # ---- Tasks ----
# --------------------------------------------------
# SERIAL EXECUTION
# --------------------------------------------------
task_research = Task(
    description=(
        "Use the web search results to explain the top 3 recent advancements in AI for healthcare in 3-4 sentences."
        "Do not call tools again after getting results."
    ),
    expected_output=(
        "Detailed notes on three advancements, with names and explanations."
    ),
    agent=researcher
)

task_write = Task(
    description=(
        "Write a short executive summary using the research notes provided by the AI Researcher. "
        "Limit the answer to about 100 words."
    ),
    expected_output=(
        "An executive summary report of the top 3 AI advancements in healthcare."
    ),
    agent=writer,
    # The context parameter explicitly instructs task_write to wait for and consume the output of task_research.
    # This enforces a linear dependency chain when the crew is run in sequential mode.
    context=[task_research]
)



crew_serial = Crew(
    agents=[researcher, writer],
    tasks=[task_research, task_write],
    # agents=[writer],
    # tasks=[task_write],
    process=Process.sequential,
    verbose=True
)


serial_result = await crew_serial.kickoff_async()
serial_end = datetime.now(timezone.utc)


print("\n[Serial Result]:\n")
try:
    print(serial_result.raw)
except AttributeError:
    print(serial_result)
```
